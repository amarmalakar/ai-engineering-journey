A DataFrame is several Series that share the same row labels.

````python
import pandas as pd
````

## Series — one column

A Series holds one list of values. I use it when I only have one kind of fact, like age.

- The **index** on the left is the row label. It starts at 0 unless I set my own.
- **`name`** is the column's name. It is not one of the values.
- `ages[0]` is the value at that label, not "the age of person number 1" in everyday counting. Label `0` is the first row.

````python
# A Series is one column. The left numbers (0, 1, 2) are the index: row labels, not the ages.
# name= is the label this column would have inside a table.
ages = pd.Series([10, 4, 12], name="Age")
print(ages)
print("column name:", ages.name)
print("first age:", ages[0])
````

## DataFrame — the whole table

A DataFrame is the full table. I build one from a dictionary:

- each **key** becomes a column name
- each **list** becomes that column's values
- every list must be the same length, because every row needs a value in every column

`data` is still a normal Python dict. `df` is the Pandas table. Those are different objects, so I read them differently.

````python
data = {
    "Name": ["Aryan", "Meera", "Kabir"],
    "Age": [10, 12, 9],
    "Favorite Game": ["Cricket", "Chess", "Football"],
    "Favorite Color": ["Red", "Blue", "Green"]
}

# Dict access: column list, then position. This is Python, not Pandas.
print("from the dict:", data["Name"][0])

df = pd.DataFrame(data)
print(df)
````

## Pulling a column back out

One column of a DataFrame is a Series. That is how the two objects connect.

- `df["Name"]` gives the Name column (a Series)
- `df.loc[0, "Name"]` gives one cell: row label 0, column Name
- `df.shape` is `(rows, columns)`
- `df.columns` is the list of column names

**Revise this:** change Kabir's age in the cell above, rerun it, then rerun the next cell. The Age column and `shape` should still make sense.

````python
names = df["Name"]
print(names)
print("\n type of one column:", type(names).__name__)
print("\n one cell:", df.loc[0, "Name"])
print("\n shape (rows, columns):", df.shape)
print("\n columns:", list(df.columns))
````
