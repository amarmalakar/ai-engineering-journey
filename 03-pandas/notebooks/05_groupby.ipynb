The line I want to remember is `df.groupby("Team")["Age"].mean()`.

`Team` is the bucket. `Age` is what I measure. `.mean()` is how I collapse the bucket. Swap any of those three and the question changes.

````python
import pandas as pd
````

## The table I practice on

Same three kids. I added `Team` because a group only teaches me something when more than one row shares the label.

Aryan and Meera are Blue. Kabir is Green. If every team name were unique, each group would be one kid and the mean would just be that kid's age.

````python
data = {
    "Name": ["Aryan", "Meera", "Kabir"],
    "Age": [10, 12, 9],
    "Team": ["Blue", "Blue", "Green"],
}
df = pd.DataFrame(data)
print(df)
````

## One number per group

`groupby` by itself is the split. Printing it shows a GroupBy object, which is the buckets before I collapse them.

Then I pick a column and one aggregation:

- `.mean()` averages the ages in each team. Blue is `(10 + 12) / 2 = 11`.
- `.count()` counts non-missing names in each team. Blue is 2, Green is 1.

The team name becomes the index of the result. There is one row per team, no longer one row per kid.

````python
grouped = df.groupby("Team")
print(grouped)

print(grouped["Age"].mean())
print(grouped["Name"].count())
````

## Several numbers at once

`.agg([...])` runs more than one summary on the same buckets. Here I want the youngest, oldest, and average age on each team.

I aggregate `Age` because it is a number. `Name` is text, so `mean` is the wrong summary for it. `count` is the summary I use on names.

````python
print(grouped["Age"].agg(["min", "max", "mean"]))
````
