````python
import pandas as pd
````

## Two tables, one shared name

Same three kids. Ages live in `df1`. Grades live in `df2`. `Name` is the connector, the column that exists in both.

Each table also has a column I do not want in the result: Team, and Subject.

````python
df1 = pd.DataFrame({
    "Name": ["Aryan", "Meera", "Kabir"],
    "Age": [10, 12, 9],
    "Team": ["Blue", "Blue", "Green"],
})
df2 = pd.DataFrame({
    "Name": ["Aryan", "Meera", "Kabir"],
    "Grade": ["A", "B", "A"],
    "Subject": ["Math", "Math", "Math"],
})
print(df1)
print(df2)
````

## Keep only Name, Age, and Grade

A merge of the full tables would also bring Team and Subject. I pass a shorter table into `merge`:

- `df1[["Name", "Age"]]` is the key plus Age
- `df2[["Name", "Grade"]]` is the key plus Grade

`on="Name"` lines the rows up by name. The result has one row per kid and only those three columns.

````python
combined = pd.merge(
    df1[["Name", "Age"]],
    df2[["Name", "Grade"]],
    on="Name",
)
print(combined)
print(list(combined.columns))
````
