## The picture I keep

| I want | I write | What I get |
| --- | --- | --- |
| Which cells are empty | `df.isnull()` | A True/False table, same shape as `df` |
| How many holes in each column | `df.isnull().sum()` | One count per column |
| Throw out rows with a hole | `df.dropna()` | A new table without those rows |
| Plug the holes | `df.fillna("Unknown")` | A new table, same rows, holes filled |

`None` in my Python dictionary shows up as `NaN` in the DataFrame. `NaN` means missing, not zero. I already saw this on the passenger with no age in the JSON note.

`dropna` and `fillna` return a new table. `df` still has the hole unless I assign the result back to `df`.

````python
import pandas as pd
````

## The table with one hole

Same three kids as the earlier notes, plus a color. Meera's color is `None`.

Pandas does not store that as the word "None". It stores `NaN`.

````python
data = {
    "Name": ["Aryan", "Meera", "Kabir"],
    "Age": [10, 12, 9],
    "Favorite Game": ["Cricket", "Chess", "Football"],
    "Favorite Color": ["Red", None, "Green"],
}
df = pd.DataFrame(data)
print(df)
````

## Find the holes

`isnull()` checks every cell. `True` means missing. `False` means the cell has a value.

Meera's color is the only `True` in this table.

`sum()` counts `True` as 1, so I get one number per column. Favorite Color should be 1. The other columns should be 0.

````python
# True where the cell is missing. Meera's color is the only True.
print(df.isnull())

# One count per column.
print(df.isnull().sum())
````

## Drop the row

`dropna()` removes any row that has at least one hole. Meera leaves. Aryan and Kabir stay, and their index labels stay 0 and 2.

I lose the whole person, not just the empty color. If I still need Meera's age and game, `dropna` is the wrong tool.

````python
# Any row with a hole is removed. Meera goes, not just her color.
clean_df = df.dropna()
print(clean_df)

# df itself still has the hole.
print(df)
````

## Fill the hole

`fillna` replaces every `NaN` and keeps the row. I use `"Unknown"` because this column is text.

A number like `0` would be a real value, and it does not belong in a color column. The first version of this cell said "a default number (like 0)" while the code filled with `"-"`. The fill has to fit the column. `"-"` and `"Unknown"` are both placeholders. `"Unknown"` is the one I want to remember.

`filled_df` is the copy. `df` still shows `NaN`.

````python
# A word, because this column is text. Not 0.
filled_df = df.fillna("Unknown")
print(filled_df)

# The original table still has NaN.
print(df)
````
