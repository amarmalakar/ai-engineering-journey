# Select, filter, and sort

I wrote this so I can reopen it later and remember three ways to look at a table: pick columns, keep some rows, or change the order.

**How I revise:** run every code cell from the top. Then change one age or one comparison (`<=` to `>`), rerun, and check who is still in the table.

## The picture I keep

| I want | I write | What I get |
| --- | --- | --- |
| One column | `df["Name"]` | A Series |
| Several columns | `df[["Name", "Age"]]` | A smaller DataFrame |
| Some rows | `df[df["Age"] <= 10]` | Only the rows that pass the test |
| A new order | `df.sort_values("Age")` | The same rows, sorted |

These calls return a new table. They do not change `df` unless I assign the result back to `df`.

````python
import pandas as pd
````

## The table I practice on

Same three kids as the Series note. I build the table here so this notebook still runs if I open it alone.

````python
data = {
    "Name": ["Aryan", "Meera", "Kabir"],
    "Age": [10, 12, 9],
    "Favorite Game": ["Cricket", "Chess", "Football"],
}
df = pd.DataFrame(data)
print(df)
````

## Select columns

`df["Name"]` pulls one column. That result is a Series, the same object from the first note.

`df[["Name", "Age"]]` pulls several columns. The inner brackets are a list of names. The result is still a DataFrame, just narrower.

````python
# One column: a Series.
print(df["Name"])

# Several columns: a list of names inside the brackets. This stays a DataFrame.
print(df[["Name", "Age"]])
````

## Filter rows

The condition goes inside the brackets: `df[df["Age"] <= 10]`. Pandas checks that test once per row and keeps the rows where it is true.

`<= 10` means 10 or younger. Aryan is 10, so he stays. Meera is 12, so she drops.

Older than 10 is a different test: `> 10`. Only Meera passes that one.

I used to label the `<= 10` line "older than 10". The symbol and the words have to match, or I will keep the wrong kids.

````python
# <= 10 keeps Age 10 and younger.
ten_or_younger = df[df["Age"] <= 10]
print(ten_or_younger)

# > 10 keeps only kids strictly older than 10.
older_than_10 = df[df["Age"] > 10]
print(older_than_10)

# The filter built new tables. df is still the original three rows.
print(df)
````

## Sort rows

`sort_values("Age")` orders smallest to biggest. The row labels travel with the person: after the sort, Kabir is still index 2.

`ascending=False` flips that, biggest first.

A list of columns sorts by the first name, then uses the next name only to break a tie. In this table every age is different, so `["Age", "Name"]` looks the same as sorting by Age alone.

**Revise this:** set Aryan's age to 9 as well, rerun, and the Name column should decide who comes first among the two 9-year-olds.

````python
# Smallest age first. Kabir stays on index 2; only the display order changes.
print(df.sort_values("Age"))

# Biggest age first.
print(df.sort_values("Age", ascending=False))

# Age first, then Name. No tied ages here, so this matches the first print.
print(df.sort_values(["Age", "Name"]))
````
