## CSV — the full table

`read_csv` reads comma-separated text. The first line of the file is the column names.

I store that table in `df`. I do not reuse `df` for the JSON file, or I lose this table.

````python
# ../data/ means: leave notebooks/, then open the file in data/.
df = pd.read_csv("../data/titanic.csv")
print("shape (rows, columns):", df.shape)
````

## First look: head, tail, shape

I do not read all 891 rows to check the load.

- `head()` is the first 5 rows. I use it to see the column names and a few passengers.
- `tail()` is the last 5 rows. On this file, that is the end of the full voyage.
- `shape` is `(rows, columns)`. This is the number I remember. The CSV has many more rows than the 6-row JSON.

````python
# First 5 rows. Enough to see the column names and a few passengers.
print(df.head())
````

````python
# Last 5 rows. These are the end of the full file, not the six-row sample.
print(df.tail())
````

````python
# (rows, columns). The CSV is the full voyage, so the row count is much larger than 6.
print(df.shape)
````

## JSON — six passengers

`sample.json` is a list of objects, one object per passenger. `read_json` turns that list into a DataFrame.

The sample is a short cut of the same voyage: 6 rows, and fewer columns than the CSV. One age is missing. Pandas shows a missing number as `NaN`, not as zero.

I store it in `df2` on purpose. If I write `df = pd.read_json(...)`, I replace the CSV table.

````python
# df2, not df, so the full CSV table stays in df.
df2 = pd.read_json("../data/sample.json")
print("json shape:", df2.shape)
print("csv shape:", df.shape)
print(df2)
print("missing age:", df2.loc[5, "Age"])
````
