| Dirt | I write | What changes |

Each of these returns a new Series or table. I assign the result back, or `df` stays messy.

````python
import pandas as pd
````

## The messy table

Same three kids, with dirt added on purpose.

- Aryan is pasted twice.
- Meera's name has a leading space.
- Team mixes `" Blue"` and `"blue"`.
- Age looks like numbers, and the column is still text.

Missing values were the last note. This note is the other kind of dirt: text that should match, rows that repeat, and numbers stored as words.

````python
data = {
    "Name": ["Aryan", " Meera", "Kabir", "Aryan"],
    "Age": ["10", "12", "9", "10"],
    "Team": [" Blue", "blue", "Green", " Blue"],
}
df = pd.DataFrame(data)
print(df)
````

## See the dirt before I fix it

`dtypes` shows Age as `str`, which means text. A mean of that column is the wrong next step.

`duplicated()` marks a row that is a copy of an earlier one. The second Aryan is `True`.

I print the Team values in a Python list so the leading space stays visible. Counting names by the raw `Team` column then splits Blue into two buckets: `" Blue"` and `"blue"`. I clean the labels before I trust a groupby.

````python
print(df.dtypes)
print(df["Name"].tolist())
print(df["Team"].tolist())
print(df.duplicated())
print(df.groupby("Team")["Name"].count())
````

## Fix text, then duplicates, then types

I do it in this order:

1. `str.strip()` removes spaces at the ends. `str.title()` turns `"blue"` into `"Blue"`.
2. `drop_duplicates()` removes the repeated Aryan row. I assign it back to `df`.
3. `astype(int)` turns the age text into integers.

After this, three kids remain. Aryan and Meera are both Blue. Kabir is Green.

````python
df["Name"] = df["Name"].str.strip()
df["Team"] = df["Team"].str.strip().str.title()
df = df.drop_duplicates()
df["Age"] = df["Age"].astype(int)
print(df)
print(df.dtypes)
````

## Check the clean table with groupby

This is the line from the groupby note. It works once `Team` is consistent and `Age` is a number.

Blue is `(10 + 12) / 2 = 11`. Green is 9.

````python
print(df.groupby("Team")["Age"].mean())
````
