# Titanic inspection

Load the raw passenger file, check columns, missing values, and groups, then write a cleaned table.

Run every code cell from the top. The kernel can be started from `03-pandas` or from `src/`.

```python
import sys
from pathlib import Path

root = Path.cwd()
src = root / "src" if (root / "src" / "clean.py").exists() else root
sys.path.insert(0, str(src))

from clean import load, inspect, clean, save
```

## Load and inspect

`load` reads `data/titanic.csv`. `inspect` prints shape, types, missing counts, duplicates, and `Age` / `Fare` summary stats.

```python
raw = load()
inspect(raw)
print(raw.head())
```

## Groups

Survival by sex, class, and port. Fare median by class. These are the numbers behind `results/summary.md`.

```python
print("survival rate:", raw["Survived"].mean())
print(raw.groupby("Sex")["Survived"].mean())
print(raw.groupby("Pclass")["Survived"].mean())
print(raw.groupby("Embarked")["Survived"].agg(["mean", "count"]))
print(raw.groupby("Pclass")["Fare"].median())
```

## Clean

Fill missing ages with the median and mark those rows in `AgeWasMissing`. Fill the two empty ports with the most common port. Replace `Cabin` with `HasCabin`, because most cabin values are empty.

The raw table stays in `raw`. The cleaned table is written to `data/processed/titanic_clean.csv`.

```python
cleaned = clean(raw)
print("missing after clean:\n", cleaned.isnull().sum())
print(cleaned[["Age", "Fare"]].describe())
save(cleaned)
```
